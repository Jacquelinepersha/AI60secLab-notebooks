A library is code someone else wrote so you do not have to. A package is how that code is organized. pip is the tool that installs it.

```python
import sys
print(sys.version)
```

```python
# See what is already installed in this environment
%pip list
```

## Installing a package
Run this in a code cell (remove the # first):

`# %pip install requests`

Then use it like this:

```python
import math

# The standard library ships with Python, no install needed
print(math.sqrt(144))
```

```python
# Third-party libraries follow the same idea: install once, import anywhere
import json
print(json.dumps({"idea": "install, import, build"}, indent=2))
```

## The rule to remember
Install with pip, import at the top of your file, and only import what you actually use.
