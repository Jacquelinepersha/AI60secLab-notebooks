# Pandas: Read Any Dataset in 3 Lines of Python

From the AI60secLab video: https://www.youtube.com/watch?v=9iFAsbPE3So

Pandas is how Python works with tables. Three lines is honestly enough to load a dataset and start learning from it.

```python
import pandas as pd
from io import StringIO

csv_data = """name,role,projects,rating
Ana,Data Scientist,12,4.8
Luis,ML Engineer,9,4.6
Priya,AI Engineer,15,4.9
Tom,Data Analyst,7,4.2
Sara,ML Engineer,11,4.7"""

df = pd.read_csv(StringIO(csv_data))  # line 1: read the data
df.head()                              # line 2: look at it
```

```python
# Line 3: describe it
df.describe(include="all")
```

```python
# Ask questions of the table
df.groupby("role")["rating"].mean()
```

```python
# Filter rows like a spreadsheet, but faster
df[df["rating"] >= 4.7]
```
