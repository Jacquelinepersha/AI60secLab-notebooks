Almost every dataset you will ever train on starts life in a database. SQL is how you ask it questions. This notebook uses SQLite, which is built into Python, so there is nothing to install.

```python
import sqlite3

conn = sqlite3.connect(":memory:")
cur = conn.cursor()

cur.execute("""CREATE TABLE lessons (
    id INTEGER PRIMARY KEY,
    title TEXT,
    subject TEXT,
    views INTEGER
)""")

rows = [
    (1, "How Does AI Learn?", "Fundamentals", 837),
    (2, "RAG in 60 Seconds", "Search", 640),
    (3, "Pandas in 3 Lines", "Python", 512),
    (4, "What Is a Neural Network?", "Deep Learning", 905),
]
cur.executemany("INSERT INTO lessons VALUES (?, ?, ?, ?)", rows)
conn.commit()
print("Table ready")
```

```python
# Ask a question: every lesson, most views first
for row in cur.execute("SELECT title, views FROM lessons ORDER BY views DESC"):
    print(row)
```

```python
# Filter with WHERE
for row in cur.execute("SELECT title FROM lessons WHERE subject = 'Python'"):
    print(row)
```

```python
# Summarize with GROUP BY
for row in cur.execute("SELECT subject, SUM(views) FROM lessons GROUP BY subject"):
    print(row)
```
